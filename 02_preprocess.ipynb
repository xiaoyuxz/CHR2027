```python
from PIL import Image
import os

folder = "../03_webtoon_raw/raw_veteran_of_the_apocalypse_episode1" 
output = "long_veteran_of_the_apocalypse_episode1.png"

files = [
    f for f in os.listdir(folder)
    if f.lower().endswith((".jpg", ".jpeg", ".png"))
]

files.sort(key=lambda x: int(os.path.splitext(x)[0]))

print("Images:", len(files))
print(files[:5])
```

```python
images = []

for f in files:
    img = Image.open(os.path.join(folder, f)).convert("RGB")
    images.append(img)

print(images[0].size)
```

```python
width = max(img.width for img in images)
height = sum(img.height for img in images)

print("Long image size:")
print(width, height)
```

```python
long_image = Image.new(
    "RGB",
    (width, height),
    "white"
)

current_height = 0

for img in images:
    long_image.paste(
        img,
        (0, current_height)
    )
    current_height += img.height
```

```python
output = "../03_webtoon_raw/long_veteran_of_the_apocalypse_episode1.png"

long_image.save(output)

print("Saved:", output)
```

#### **2. Sliding window generation**

```python
from PIL import Image
from pathlib import Path


# Allow extremely large webtoon images

Image.MAX_IMAGE_PIXELS = None


# Input

long_image_path = Path(
    "../03_webtoon_raw/long_veteran_of_the_apocalypse_episode1.png"
)


# sliding window
output_dir = Path(
    "../03_webtoon_raw/window_veteran_of_the_apocalypse_episode1"
)


output_dir.mkdir(
    parents=True,
    exist_ok=True
)



# 2. Parameters

window_height = 3000
overlap = 500

stride = window_height - overlap



# 3. Open image

img = Image.open(long_image_path).convert("RGB")

width, height = img.size


print("Original size:")
print("width:", width)
print("height:", height)


# 4. Sliding window

window_id = 1

y = 0


while y < height:

    # bottom boundary
    y_end = min(
        y + window_height,
        height
    )


    # crop window
    crop = img.crop(
        (
            0,
            y,
            width,
            y_end
        )
    )


    # filename
    filename = (
        f"window_{window_id:03d}.png"
    )


    save_path = output_dir / filename


    crop.save(
        save_path
    )


    print(
        f"Saved {filename}: y={y}-{y_end}"
    )


    window_id += 1


    # move window
    y += stride


    # stop condition
    if y >= height:
        break



print("===================")
print("Finished!")
print("Total windows:", window_id-1)
print("Saved to:", output_dir)
```
