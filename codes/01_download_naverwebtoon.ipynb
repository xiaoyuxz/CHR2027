```python
import re

html = r.text

# 找所有 URL
urls = re.findall(r'https?://[^"\']+', html)

# 筛选 comic 图片
img_urls = [
    u for u in urls 
    if "image-comic.pstatic.net" in u
]

print(len(img_urls))

for u in img_urls[:10]:
    print(u)
```

```python
import requests

url = img_urls[0]

headers = {
    "User-Agent": "Mozilla/5.0",
    "Referer": "https://comic.naver.com/"
}

r = requests.get(
    url,
    headers=headers,
    timeout=20
)

print(r.status_code)
print(len(r.content))
```

```python
import os
import time
import requests

os.makedirs("./episode1", exist_ok=True)

headers = {
    "User-Agent": "Mozilla/5.0",
    "Referer": "https://comic.naver.com/"
}

session = requests.Session()
session.headers.update(headers)


for i, url in enumerate(img_urls):

    try:
        r = session.get(
            url,
            timeout=30
        )

        r.raise_for_status()

        with open(f"./episode1/{i:03d}.jpg", "wb") as f:
            f.write(r.content)

        print(i, "done")

        time.sleep(0.5)

    except Exception as e:
        print(i, "failed", e)
```
